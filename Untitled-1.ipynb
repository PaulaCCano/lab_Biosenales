```python
import numpy as np
import mne
import os
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
```

```python
import os

# Directorio base donde tienes la carpeta 'files'
base_dir = r"D:\LabBiosenales\data_prueba"

# Listas para almacenar las rutas según la tarea
rutas_reposo = []
rutas_movimiento = []
rutas_imagineria = []

for i in range(1, 110):  # Del sujeto 1 al 109
    sujeto = f"S{i:03d}"  # Formatea a S001, S002, etc.
    carpeta_sujeto = os.path.join(base_dir, sujeto)
    
    # Rutas para Reposo (R01, R02)
    rutas_reposo.append(os.path.join(carpeta_sujeto, f"{sujeto}R01.edf"))
    
    # Rutas para Movimiento Real (R03, R07, R11)
    for run in ["R03", "R07", "R11"]:
        rutas_movimiento.append(os.path.join(carpeta_sujeto, f"{sujeto}{run}.edf"))
        
    # Rutas para Imaginería (R04, R08, R12)
    for run in ["R04", "R08", "R12"]:
        rutas_imagineria.append(os.path.join(carpeta_sujeto, f"{sujeto}{run}.edf"))
```

```python
def calcular_potencia_banda(epocas, fmin=8.0, fmax=30):
    """
    Calcula la Densidad Espectral de Potencia (PSD) en una banda específica.
    """
    # MNE tiene una función integrada para calcular PSD
    psd = epocas.compute_psd(method='welch', fmin=fmin, fmax=fmax, verbose=False)
    
    # psd.get_data() devuelve una matriz: (épocas, canales, frecuencias)
    datos_psd = psd.get_data()
    
    # 1. Promediamos a través del eje de las frecuencias (para tener un solo valor de la banda Mu)
    potencia_mu = np.mean(datos_psd, axis=2)
    
    # 2. Promediamos a través de todas las épocas para tener un valor por canal
    potencia_promedio_sujeto = np.mean(potencia_mu, axis=0)
    
    return potencia_promedio_sujeto
```

```python
def procesar_archivo_edf(ruta_archivo, tipo_tarea, tmax=4.0):
    """
    Carga un archivo .edf y extrae las épocas dependiendo de si es Reposo o Tarea (Movimiento/Imaginería).
    """
    # 1. Leer el archivo EDF
    raw = mne.io.read_raw_edf(ruta_archivo, preload=True, verbose=False)
    
    # Estandarizar nombres de canales (quitar puntos o espacios si los hay en el EDF)
    raw.rename_channels(lambda x: x.strip('.'))
    raw.pick(['eeg'])
    
    # Filtro enfocado en tu banda de interés (Mu)
    raw.filter(9, 30, method='fir', fir_design='firwin', verbose=False)
    
    if tipo_tarea == 'reposo':
        # Como no hay eventos T1/T2, creamos épocas fijas de 4 segundos
        epocas = mne.make_fixed_length_epochs(raw, duration=tmax, preload=True, verbose=False)
        return epocas, None # Retornamos None en el segundo valor porque no hay Izquierda/Derecha
        
    else:
        # Extraemos eventos T1 (Izquierda) y T2 (Derecha) de los archivos EDF
        eventos, id_eventos = mne.events_from_annotations(raw, verbose=False)
        
        # En los EDF crudos de PhysioNet, los eventos suelen llamarse 'T1' y 'T2'
        # Filtramos el diccionario id_eventos para quedarnos solo con esos
        eventos_validos = {k: v for k, v in id_eventos.items() if k in ['T1', 'T2']}
        
        epocas = mne.Epochs(raw, eventos, event_id=eventos_validos, tmin=0, tmax=tmax, 
                            baseline=None, preload=True, verbose=False)
        
        # Separar por lateralidad
        epocas_izq = epocas['T1']
        epocas_der = epocas['T2']
        
        return epocas_izq, epocas_der
```

```python
epocas_izq, epocas_der = procesar_archivo_edf(rutas_movimiento[0], 'movimiento')
potencia_izq = calcular_potencia_banda(epocas_izq)
potencia_der = calcular_potencia_banda(epocas_der)
print("Potencia en banda Mu para Movimiento Izquierda:", potencia_izq)
print("Potencia en banda Mu para Movimiento Derecha:", potencia_der)
```

```python
# gráfico del espectro de potencia para el primer sujeto y la primera tarea de movimiento
plt.figure(figsize=(10, 5))
sns.barplot(x=np.arange(len(potencia_izq)), y=potencia_izq, color='blue', alpha=0.6, label='Movimiento Izquierda')
sns.barplot(x=np.arange(len(potencia_der)), y=potencia_der, color='red', alpha=0.6, label='Movimiento Derecha')
plt.xlabel('Canales EEG')
plt.ylabel('Potencia en banda Mu')
plt.title('Potencia en banda Mu para Movimiento Izquierda vs Derecha (Sujeto 1)')
plt.legend()
```
